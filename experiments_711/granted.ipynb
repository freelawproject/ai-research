# Parse granted list and associated reports for lower court, lower court case name, and lower court decision citation

SCOTUS produces [granted list](https://www.supremecourt.gov/orders/grantednotedlists.aspx) for cases granted for arguments, this list shows the lower court, dates, as well as the decision (when available), and for some cases, they also produce a report (eg., [24-568](https://www.supremecourt.gov/qp/24-00568qp.pdf)) which also shows the lower court decision citation and the questions presented.

We can use this granted list to get the lower court information and fill in data gaps. This notebook produces the granted.csv report as of 7/17/2025, containing information since 2007. This list is updated periodically and for 2024 and 2025 granted list, we should download and refresh the table periodically to get the latest information, until the list is finalized for that year.
