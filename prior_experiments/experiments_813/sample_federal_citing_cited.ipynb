# Load the citing data extracted from CLReplica

```python
with open('data/fed/citing_opinions.json', 'r') as f:
    results = json.load(f)
```

```python
citing_metadata = pd.DataFrame.from_dict(results, orient='index')
citing_metadata = citing_metadata.reset_index().rename(columns={'index': 'citing_cluster_id'})
citing_metadata.head()
```

```python
citing_metadata.columns
```

```python
citing_metadata = citing_metadata[['citing_cluster_id', 'citing_url', 'citing_court_name', 'cited_cluster_ids']]
citing_exploded = citing_metadata.explode("cited_cluster_ids", ignore_index=True)

citing_exploded.rename(columns={"cited_cluster_ids": "cited_cluster_id"}, inplace=True)
citing_exploded["cited_cluster_id"] = citing_exploded["cited_cluster_id"].astype(int)
citing_exploded.head()
```

# Load the cited data extracted from CLReplica

```python
with open('data/fed/cited_opinions.json', 'r') as f:
    results = json.load(f)
```

```python
cited_metadata = pd.DataFrame.from_dict(results, orient='index')
cited_metadata = cited_metadata.reset_index().rename(columns={'index': 'cited_cluster_id'})
cited_metadata.head()
```

```python
cited_metadata.columns
```

```python
cited_metadata = cited_metadata[['cited_cluster_id', 'cited_url', 'cited_court_name', 'cited_case_name_short', 'cited_case_name', 'cited_citations']]
cited_metadata["cited_cluster_id"] = cited_metadata["cited_cluster_id"].astype(int)
cited_metadata.head()
```

# Create the output df

```python
output = pd.merge(citing_exploded, cited_metadata, how="left", on="cited_cluster_id")
output.head()
```

```python
len(output)
```

```python
for col in eda_cols[:-1]:
    print("----------")
    print(output[col].nunique())
    display(output[col].value_counts())
```

# Save the output df

```python
assert output["cited_cluster_id"].isnull().sum() == 0
```

```python
assert output["cited_url"].isnull().sum() == 0
```

```python
output.to_csv("data/fed/federal_citing_cited.csv", index=False)
```

# Save each citing cluster as a separate csv for expert annotation

```python
for citing_cluster_id in output["citing_cluster_id"].unique():
    annotate = output[output["citing_cluster_id"] == citing_cluster_id].copy()
    annotate["expert"] = None
    annotate["expert_label"] = None
    annotate["expert_notes"] = None
    num_cited = annotate["cited_cluster_id"].nunique()

    expert_folder = f"data/fed/annotations/"
    
    annotate.to_csv(f"{expert_folder}/{num_cited}_{citing_cluster_id}.csv", index=False)
```
