# Clean up output.json

Bill helped me query the database to get all subsequent opinions that cited the cited_cluster_id opinions. This notebook outlines all the steps I undertook to clean the output and filter for only SCOTUS citing opinions.

# Import library

```python
import numpy as np
import pandas as pd

import json
```

# Load the data & clean

```python
with open("citing/output.json") as f:
    data = json.load(f)

# Flatten into list of rows
rows = []
for outer_id, items in data.items():
    for item in items:
        item["cited_cluster_id"] = int(outer_id)  # Add the outer key as a new column
        rows.append(item)

# Create DataFrame
df = pd.DataFrame(rows)
df.head()
```

```python
len(df)
```

```python
df = df.rename(columns={'court': 'citing_court_id', 'id': 'citing_cluster_id'})
df = df[["cited_cluster_id", "citing_cluster_id", "citing_court_id"]]
df.head()
```

# Save cleaned output for future use

```python
df.to_csv("citing/clean_output.csv", index=False)
```

# Filter for SCOTUS & save for future use

```python
scotus = df[df["citing_court_id"] == "scotus"]
len(scotus)
```

```python
scotus.to_csv("citing/scotus.csv", index=False)
```
